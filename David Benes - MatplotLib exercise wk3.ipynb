```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\benes\OneDrive\Desktop\School\machine learning\car data.csv")
df
```

# 3 Barchart

```python
years = df["Year"].value_counts()
years.plot(kind="bar")
plt.title("Number of cars by year")
plt.tight_layout()
plt.show()
```

# 4 Histogram

```python
plt.hist(df["Selling_Price"], bins=20, edgecolor = "black")
plt.title("Selling price")
plt.tight_layout()
plt.show()
```

# 5 piechart

```python
fuel_counts = df["Fuel_Type"].value_counts()
plt.pie(fuel_counts, labels=fuel_counts.index)
plt.title("Cars by Fuel Type")
plt.tight_layout()
plt.show()
```
